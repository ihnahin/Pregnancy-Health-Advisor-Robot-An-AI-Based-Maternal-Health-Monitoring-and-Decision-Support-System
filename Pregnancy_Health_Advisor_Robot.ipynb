```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display, Markdown
```

```python
file_path = "/content/Maternal Health Risk Data Set.csv"
df = pd.read_csv(file_path)

print("✅ Dataset Loaded Successfully!\n")
print("📊 Dataset Preview:")
display(df.head())

print("\n📈 Dataset Summary:")
display(df.describe())
```

```python
print("👉 Please enter patient details:\n")

patient = {
    "Name": input("Patient Name: "),
    "Blood Group": input("Blood Group: "),
    "Pregnancy Month": int(input("Pregnancy Month (1-9): ")),
    "Blood Pressure": int(input("Blood Pressure (Systolic): ")),
    "Hemoglobin": float(input("Hemoglobin Level (g/dl): ")),
    "Sugar Level": int(input("Sugar Level (mg/dl): ")),
    "Estimated Baby Weight": float(input("Estimated Baby Weight (kg): "))
}
```

```python
paragraph_report = []
point_guidelines = []

if patient["Blood Pressure"] > 140:
    paragraph_report.append(
        f"⚠️ Your blood pressure is higher than the normal range (above 140 mmHg). "
        f"This condition may indicate risk of pre-eclampsia, a dangerous complication during pregnancy. "
        f"You should reduce salty and oily food, stay hydrated, and check your blood pressure daily. "
        f"If it remains high, consult your doctor immediately."
    )
    point_guidelines += ["➡️ Monitor BP daily","➡️ Reduce salt and oily foods","❌ Avoid stress and junk food"]

elif patient["Blood Pressure"] < 90:
    paragraph_report.append(
        f"⚠️ Your blood pressure is lower than normal (below 90 mmHg). "
        f"This can cause dizziness, weakness, and fainting. "
        f"Increase fluid intake, take proper rest, and avoid sudden standing. "
        f"If symptoms persist, seek medical advice."
    )
    point_guidelines += ["➡️ Drink fluids","➡️ Take rest","❌ Avoid sudden standing"]

else:
    paragraph_report.append(
        f"✅ Your blood pressure is within the healthy range (90–140 mmHg). "
        f"Maintain a balanced diet, regular rest, and light activity to keep it stable."
    )


if patient["Hemoglobin"] < 11:
    paragraph_report.append(
        f"⚠️ Your hemoglobin level ({patient['Hemoglobin']} g/dl) is below the recommended level (11 g/dl). "
        f"This indicates a risk of anemia, which may cause weakness, fatigue, and complications during delivery. "
        f"Consume iron-rich foods like leafy vegetables, dates, pomegranate, and consider iron supplements as prescribed."
    )
    point_guidelines += ["➡️ Eat leafy vegetables, dates, pomegranate","➡️ Take iron supplements","❌ Avoid tea/coffee after meals"]

else:
    paragraph_report.append(
        f"✅ Your hemoglobin level ({patient['Hemoglobin']} g/dl) is within the normal range. "
        f"Continue a balanced diet to maintain it."
    )


if patient["Sugar Level"] > 140:
    paragraph_report.append(
        f"⚠️ Your blood sugar level ({patient['Sugar Level']} mg/dl) is higher than normal. "
        f"This indicates a possibility of gestational diabetes. "
        f"You should avoid sugary foods, eat small frequent meals, and check blood sugar regularly. "
        f"Consult your doctor for further testing."
    )
    point_guidelines += ["➡️ Monitor sugar","➡️ Eat small frequent meals","❌ Avoid sweets and soft drinks"]

elif patient["Sugar Level"] < 70:
    paragraph_report.append(
        f"⚠️ Your blood sugar level ({patient['Sugar Level']} mg/dl) is lower than normal (below 70 mg/dl). "
        f"This can cause weakness, dizziness, and risk of fainting. "
        f"Eat snacks frequently, carry glucose or biscuits, and never skip meals."
    )
    point_guidelines += ["➡️ Eat snacks frequently","➡️ Carry glucose/biscuits","❌ Don’t skip meals"]

else:
    paragraph_report.append(
        f"✅ Your blood sugar level ({patient['Sugar Level']} mg/dl) is within the normal range (70–140 mg/dl). "
        f"Maintain a balanced diet and regular checkups."
    )


if patient["Estimated Baby Weight"] < 2.3 and patient["Pregnancy Month"] >= 7:
    paragraph_report.append(
        f"⚠️ The baby’s estimated weight ({patient['Estimated Baby Weight']} kg) seems lower than expected for month {patient['Pregnancy Month']}. "
        f"This may indicate growth restriction. Ensure you consume protein-rich foods (milk, eggs, pulses) and rest adequately. "
        f"A follow-up with your doctor is recommended."
    )
    point_guidelines += ["➡️ Eat eggs, milk, pulses","➡️ Rest properly","❌ Don’t skip meals"]

elif patient["Estimated Baby Weight"] > 3.5:
    paragraph_report.append(
        f"⚠️ The baby’s estimated weight ({patient['Estimated Baby Weight']} kg) is higher than average. "
        f"This can cause complications during delivery. Maintain a balanced diet, avoid overeating carbs, and engage in light activity if allowed by your doctor."
    )
    point_guidelines += ["➡️ Avoid excess carbs","➡️ Walk daily","❌ Don’t sit idle all day"]

else:
    paragraph_report.append(
        f"✅ The baby’s weight ({patient['Estimated Baby Weight']} kg) is within the healthy range for month {patient['Pregnancy Month']}."
    )
```

```python
report_text = f"""
# 🧾 Pregnancy Health Advisor Robot


**👩 Patient Name:** {patient['Name']}
**🩸 Blood Group:** {patient['Blood Group']}
**📅 Pregnancy Month:** {patient['Pregnancy Month']}
**💓 Blood Pressure:** {patient['Blood Pressure']} mmHg
**🧬 Hemoglobin:** {patient['Hemoglobin']} g/dl
**🍬 Sugar Level:** {patient['Sugar Level']} mg/dl
**👶 Estimated Baby Weight:** {patient['Estimated Baby Weight']} kg

---

## 📖 Detailed Report:
"""

for para in paragraph_report:
    report_text += f"- {para}\n"

report_text += "\n---\n## 📌 Guidelines:\n"
for g in point_guidelines:
    report_text += f"{g}\n"

display(Markdown(report_text))
```

```python
plt.figure(figsize=(7,5))
metrics = ["Blood Pressure", "Hemoglobin", "Sugar Level", "Baby Weight"]
values = [patient["Blood Pressure"], patient["Hemoglobin"], patient["Sugar Level"], patient["Estimated Baby Weight"]]

plt.bar(metrics, values, color=['orange','green','red','blue'])
plt.title("📊 Patient Health Overview")
plt.ylabel("Values")
plt.show()
```

```python
plt.figure(figsize=(7,4))
plt.axhline(90, color='green', linestyle='--', label="Normal Min (90)")
plt.axhline(140, color='red', linestyle='--', label="Normal Max (140)")
plt.plot([1], [patient["Blood Pressure"]], marker='o', markersize=10, color='orange', label="Patient BP")

plt.xticks([])
plt.ylabel("Blood Pressure (mmHg)")
plt.title("💓 Blood Pressure vs Normal Range")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.bar(["Patient Sugar"], [patient["Sugar Level"]],
        color="red" if patient["Sugar Level"]>140 or patient["Sugar Level"]<70 else "green")

plt.axhline(70, color="blue", linestyle="--", label="Low Threshold")
plt.axhline(140, color="orange", linestyle="--", label="High Threshold")

plt.ylabel("mg/dl")
plt.title("🍬 Sugar Level Status")
plt.legend()
plt.show()
```

```python
months = np.arange(1,10)
ideal_weight = [0.1,0.3,0.6,1.0,1.5,2.0,2.5,3.0,3.3]  # Ideal curve

plt.figure(figsize=(7,5))
plt.plot(months, ideal_weight, label="Ideal Growth Curve", color="green")
plt.scatter(patient["Pregnancy Month"], patient["Estimated Baby Weight"],
            color="red", s=100, label="Patient Baby Weight")

plt.xlabel("Pregnancy Month")
plt.ylabel("Baby Weight (kg)")
plt.title("👶 Baby Weight Growth Tracker")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
